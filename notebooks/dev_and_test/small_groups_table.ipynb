# Small Groups of Order 1–20: A Reference Table

A generic, implementation-independent catalogue of every group of order $n$ for $1\le n\le 20$
(54 groups in total: 31 abelian, 23 non-abelian). For each order there is one section with one row per
group (up to isomorphism), giving standard and alternative notation, standard and alternative names,
and the construction needed to build the group from simpler pieces.

This notebook is a planning aid for building a list of sublists, one sublist per order, and is deliberately
not tied to any particular software library.

## Conventions

**Notation**

| Symbol | Meaning |
|---|---|
| $Z_n$ (alt. $C_n$, $\mathbb{Z}/n\mathbb{Z}$) | cyclic group of order $n$ |
| $D_n$ (alt. $\mathrm{Dih}_n$) | dihedral group of the regular $n$-gon, of order $2n$. Many texts write $D_{2n}$ for the same group, so that spelling is listed as an alternative. |
| $\mathrm{Dic}_n$ (alt. $\mathrm{BD}_{4n}$, $2D_n$) | dicyclic (binary dihedral) group of order $4n$, with presentation $\langle a,x \mid a^{2n}=e,\ x^2=a^n,\ x^{-1}ax=a^{-1}\rangle$. $\mathrm{Dic}_2=Q_8$, $\mathrm{Dic}_4=Q_{16}$. |
| $V_4$ | Klein four-group, $\cong Z_2\times Z_2$ |
| $S_n$, $A_n$ | symmetric and alternating groups on $n$ letters |
| $Z_n^k$ | direct product of $k$ copies of $Z_n$ |

**Construction operations**

| Symbol | Operation | Meaning |
|---|---|---|
| $\times$ | direct product | $G\times H$: componentwise multiplication |
| $\rtimes$ | semidirect product | $N\rtimes_\varphi H$: $H$ acts on the normal subgroup $N$ through a homomorphism $\varphi:H\to\mathrm{Aut}(N)$ |
| $\circ$ | central product | $G\circ H$: take $G\times H$ and identify a central subgroup of $G$ with a central subgroup of $H$ |
| non-split extension | | the group cannot be written as a (semi)direct product of proper subgroups, e.g. $Q_8$, $Q_{16}$ (the dicyclic groups $\mathrm{Dic}_2$, $\mathrm{Dic}_4$) |

**Columns.** *Operation* names the primary construction a program would need. A group often has several
valid descriptions (for instance $Z_6\cong Z_2\times Z_3$); the *Construction* column gives the most useful
one and mentions equivalent forms. *SmallGroup ID* is the pair `(order, index)` in the GAP SmallGroups library,
included only as a cross-reference; the index is not used anywhere else in this notebook.

## Notes for the next step

- The **Operation** column indicates which constructors are needed to produce all 54 groups:
  cyclic groups, direct products ($\times$), semidirect products ($\rtimes$, with the action specified),
  one central product ($\circ$, the Pauli group at order 16), and the dicyclic groups
  $Q_8$ and $Q_{16}$, which are non-split extensions and need a presentation or explicit multiplication rule.
- Every non-abelian group of order up to 20 except $Q_8$ and $Q_{16}$ is a semidirect product (a direct
  product being the case of trivial action). The Pauli group is also $(Z_4\times Z_2)\rtimes Z_2$, but the
  central product $Z_4\circ D_4$ is its most natural description.
- Distinct groups can share a written form: $\mathrm{Dic}_5$ and $F_{20}$ are both $Z_5\rtimes Z_4$,
  and $D_4$, $M_{16}$, $D_8$, $SD_{16}$ all have a $Z_n\rtimes Z_2$ form. The action is what distinguishes them.
