# Groups of order 13 through 20

30 groups, one entry each, in the order of GAP's SmallGroup library. An entry gives the group's SmallGroup id `(order, index)`, its finalg name, some facts computed with GAP, a short note, and the finalg code that builds it.

* The cells are independent of each other, except that they all use the first (setup) cell.
* Each code cell ends by showing `(name, order, is_commutative)`.
* Every group was checked by exporting its Cayley table to GAP and comparing `IdGroup` with the id in the heading.
* In the facts, |Z(G)| is the order of the centre, |G′| the order of the derived subgroup, and `n (×k)` means that there are k elements of order n.
* Names: `Zn` is cyclic of order n; `Dn` is dihedral of order 2n; `Dicn`, `Q8`, `Q16`, `Q32` are dicyclic and generalized quaternion groups; `x` means direct product and `sdp` semi-direct product (a trailing `_kK` or `_SGn_i` tells apart different groups with the same factors).
* JSON files are in the `json` folder; `load_group` and `load_groups` in `small_groups_tools.py` read them (see README).

## Setup

```python
from finalg import (small_groups, generate_cyclic_group, generate_dihedral_group,
                    generate_dicyclic_group, generate_alternating_group)
from small_groups_tools import mult_action

def z(n):
    """The cyclic group of order n."""
    return generate_cyclic_group(n)

def dihedral(n):
    """The dihedral group D_n, of order 2n."""
    return generate_dihedral_group(n)[0]

def dicyclic(n):
    """The dicyclic group Dic_n, of order 4n."""
    return generate_dicyclic_group(n)[0]
```

### SmallGroup(16, 3) · `Z2Z2_sdp_Z4`

**GAP structure:** (C4 x C2) : C2 · **order** 16 · **non-abelian**  
|Z(G)| = 4 · |G′| = 2 · 10 conjugacy classes · exponent 4  
**Element orders:** 1 (×1), 2 (×7), 4 (×8)

(Z2 × Z2) ⋊ Z4, where the generator of Z4 swaps the two copies of Z2.

Built with `Group.semidirect_product`. In finalg it is `small_groups(16)[2]`.

**File:** `json/G16_03_Z2Z2_sdp_Z4.json`

```python
swap = lambda h, x: x if int(h) % 2 == 0 else ':'.join(reversed(x.split(':')))
G = (z(2) * z(2)).semidirect_product(z(4), swap, name='Z2Z2_sdp_Z4')
(G.name, G.order, G.is_commutative())
```

### SmallGroup(16, 13) · `Pauli`

**GAP structure:** (C4 x C2) : C2 · **order** 16 · **non-abelian**  
|Z(G)| = 4 · |G′| = 2 · 10 conjugacy classes · exponent 4  
**Element orders:** 1 (×1), 2 (×7), 4 (×8)

The Pauli group generated by the Pauli matrices: the central product Z4 ∘ D4 of order 16, formed by identifying the central involution of D4 with the element of order 2 in Z4. It is also Z4 ∘ Q8.

Built with `Group.central_product`. In finalg it is `small_groups(16)[12]`.

**File:** `json/G16_13_Pauli.json`

```python
d4 = dihedral(4)
central_involution = [c for c in d4.center() if c != d4.identity][0]
G = z(4).central_product(d4, ('2', central_involution), name='Pauli')
(G.name, G.order, G.is_commutative())
```
