# Groups of order 33 through 40

37 groups, one entry each, in the order of GAP's SmallGroup library. An entry gives the group's SmallGroup id `(order, index)`, its finalg name, some facts computed with GAP, a short note, and the finalg code that builds it.

* The cells are independent of each other, except that they all use the first (setup) cell.
* Each code cell ends by showing `(name, order, is_commutative)`.
* Every group was checked by exporting its Cayley table to GAP and comparing `IdGroup` with the id in the heading.
* In the facts, |Z(G)| is the order of the centre, |G′| the order of the derived subgroup, and `n (×k)` means that there are k elements of order n.
* Names: `Zn` is cyclic of order n; `Dn` is dihedral of order 2n; `Dicn`, `Q8`, `Q16`, `Q32` are dicyclic and generalized quaternion groups; `x` means direct product and `sdp` semi-direct product (a trailing `_kK` or `_SGn_i` tells apart different groups with the same factors).
* The JSON files are in `src/finalg/data/small_groups/json` (the **File** lines below are relative to `src/finalg/data/small_groups`); `load_group` and `load_groups` in `finalg.small_algebras` read them (see README).

## Setup

```python
from finalg import (small_groups, generate_cyclic_group, generate_dihedral_group,
                    generate_dicyclic_group, generate_symmetric_group)
from finalg.utilities import mult_action, action_from_generators

def z(n):
    """The cyclic group of order n."""
    return generate_cyclic_group(n)

def dihedral(n):
    """The dihedral group D_n, of order 2n."""
    return generate_dihedral_group(n)[0]

def dicyclic(n):
    """The dicyclic group Dic_n, of order 4n."""
    return generate_dicyclic_group(n)[0]

# Building blocks: every group of order 20 or less, by name, from finalg's own catalogue
small = {g.name: g for grps in small_groups().values() for g in grps}
```

**GAP structure:** (C2 x C2) : C9 · **order** 36 · **non-abelian**  
|Z(G)| = 3 · |G′| = 4 · 12 conjugacy classes · exponent 18  
**Element orders:** 1 (×1), 2 (×3), 3 (×2), 6 (×6), 9 (×24)
