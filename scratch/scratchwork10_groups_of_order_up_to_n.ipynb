# Finite Groups of Order up to N

```python
import finalg as fg
```

```python
fg.examples.about()
```

```python
def z(n):
    """Return the cyclic group of order n, where n is a positive integer."""
    return fg.generate_cyclic_group(n)
```

```python
s3 = fg.generate_symmetric_group(3)
d4, _, _ = fg.generate_dihedral_group(4)
d5, _, _ = fg.generate_dihedral_group(5)
d6, _, _ = fg.generate_dihedral_group(6)
a4 = fg.generate_alternating_group(4)
q8 = fg.examples[13]
```

```python
groups = { 1: [z(1)],
           2: [z(2)],
           3: [z(3)],
           4: [z(4), z(2)*z(2)],
           5: [z(5)],
           6: [z(6), s3],
           7: [z(7)],
           8: [z(8), z(4)*z(2), z(2)**3, d4, q8],
           9: [z(9), z(3)*z(3)],
           10: [z(10), d5],
           11: [z(11)],
           12: [z(12), z(6)*z(2), a4, d6]  # INCOMPLETE: Needs the Dicyclic / Binary dihedral group
         }
```

```python
len(groups)
for ord in range(1, len(groups) + 1):
    print()
    for g in groups[ord]:
        print(f"  {g.name} - {g.order}")
```
