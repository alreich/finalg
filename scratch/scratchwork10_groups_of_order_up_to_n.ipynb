# Finite Groups of Order up to N

```python
import finalg as fg
```

```python
fg.examples.about()
```

```python
def z(n):
    """Return the cyclic group of order n, where n is a positive integer."""
    return fg.generate_cyclic_group(n)
```

```python
s3 = fg.generate_symmetric_group(3)
d4, _, _ = fg.generate_dihedral_group(4)
d5, _, _ = fg.generate_dihedral_group(5)
d6, _, _ = fg.generate_dihedral_group(6)
d7, _, _ = fg.generate_dihedral_group(7)
d8, _, _ = fg.generate_dihedral_group(8)
d9, _, _ = fg.generate_dihedral_group(9)
d10, _, _ = fg.generate_dihedral_group(10)
a4 = fg.generate_alternating_group(4)
# q8 = fg.examples[13]
dic2, _, _ = fg.generate_dicyclic_group(2)  # Isomorphic to q8
dic3, _, _ = fg.generate_dicyclic_group(3)  # Order 4*3 = 12
dic4, _, _ = fg.generate_dicyclic_group(4)
dic5, _, _ = fg.generate_dicyclic_group(5)
```

```python
groups = { 1: [z(1)],
           2: [z(2)],
           3: [z(3)],
           4: [z(4), z(2)*z(2)],
           5: [z(5)],
           6: [z(6), s3],
           7: [z(7)],
           8: [z(8), z(4)*z(2), z(2)**3, d4, dic2],
           9: [z(9), z(3)*z(3)],
           10: [z(10), d5],
           11: [z(11)],
           12: [z(12), z(6)*z(2), a4, d6, dic3],
           13: [z(13)],
           14: [z(14), d7],
           15: [z(15)],
           16: [z(16), z(8)*z(2), z(4)*z(4), z(4)*z(2)*z(2), z(2)**4, d8, dic4], # Needs 7 more
           17: [z(17)],
           18: [z(18), z(6)*z(3), d9, s3*z(3)], # Needs Generalized Dihedral (Z3*Z3) X| Z2
           19: [z(19)],
           20: [z(20), z(10)*z(2), d10, dic5] # Needs Frobenius group F20 = Z5 X| Z4
         }
```

```python
for ord, grps in groups.items():
    print()
    for g in grps:
        print(f"  {g.name} - {g.order}")
```
