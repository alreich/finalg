How should that call to `semidirect_product` be written?  (NOTE: $Z_2 * Z_2$ is the Klein-4 group.)

# Claude

````python
Z2 = fg.generate_cyclic_group(2)
Z4 = fg.generate_cyclic_group(4)
K4 = Z2 * Z2                          # elements '0:0', '0:1', '1:0', '1:1'
d = K4.direct_product_delimiter()     # ':' by default

# h acts as (swap)^h, so even h acts trivially and odd h swaps (a,b) -> (b,a)
swap = lambda h, x: x if int(h) % 2 == 0 else d.join(reversed(x.split(d)))

G = K4.semidirect_product(Z4, swap)

G.about(max_size=16)
````

The call is below. The only twist is that the elements of `Z2 * Z2` are strings like `'0:1'`, so the action has to split each name on the delimiter to swap the two coordinates. I ran it.

```python
Z2 = generate_cyclic_group(2)
Z4 = generate_cyclic_group(4)

```

One thing to settle first is the interface. A central product of G and H needs a central subgroup of each and an isomorphism between them, which is the identification.
