Here is the call. I ran it against your current `finalg` plus the new method:

```python
from finalg import generate_cyclic_group

Z8 = generate_cyclic_group(8)   # elements '0', '1', ..., '7'
Z2 = generate_cyclic_group(2)   # elements '0', '1'

# The order-2 element '1' acts on Z8 by x -> x^(-1), i.e. x -> -x (mod 8).
# The identity '0' acts trivially.
G = Z8.semidirect_product(Z2, lambda h, x: str((int(x) * (-1) ** int(h)) % 8))
```

The action is a function `action(h, x)`. Elements of `Z_n` are strings, so the lambda converts them to ints and back. For `h = '0'` the factor `(-1)**0` is 1, which is the identity map. For `h = '1'` it is -1, which inverts, since the group operation on `Z_8` is addition mod 8.

The result is `Z8_sdp_Z2`, a `Group` of order 16 with elements named `'0:0'`, `'0:1'`, `'1:0'`, and so on. It is non-abelian, with element orders 1 (×1), 2 (×9), 4 (×2) and 8 (×4). That matches the dihedral group of order 16, D16 (D8 in the other convention, the symmetries of an octagon): one rotation subgroup of order 8, plus 8 reflections and a central rotation by 180°, giving 9 involutions. The defining relation also holds: `G.op('0:1', '1:0')` is `'7:1'` and `G.op('1:0', '0:1')` is `'1:1'`, so conjugating the generator `1:0` by `0:1` inverts it.

I did not call `isomorphic` against `generate_dihedral_group(8)`. It brute-forces mappings and ran out of memory at order 12 earlier, so it won't be practical at order 16.
