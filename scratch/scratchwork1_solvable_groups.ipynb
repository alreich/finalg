```python
def solvable(grp):
    """Return True or False depending on whether the group is solvable.

    If successive applications of the commutator_subalgebra() method leads
    to the trivial group, then it's solvable. Otherwise, it's not solvable.
    """

    def solvable_aux(grp, sub):
        result = True
        if grp.order > sub.order:
            if sub.order == 1:
                result = True
            else:
                result = solvable_aux(sub, sub.commutator_subalgebra())
        else:
            result = False
        return result

    return solvable_aux(grp, grp.commutator_subalgebra())
```

```python
def solvable(grp):
    """Return True or False depending on whether the group is solvable.
    If successive applications of the commutator_subalgebra() method leads
    to the trivial group, then it's solvable. Otherwise, it's not solvable.
    """
    def solvable_aux(grp, sub):
        if sub.order == 1:
            return True
        elif sub.order == grp.order:
            return False
        else:
            return solvable_aux(sub, sub.commutator_subalgebra())
            
    return solvable_aux(grp, grp.commutator_subalgebra())
```
