# The Binary Tetrahedral Group

## Summary

This notebook constructively demonstrates that the 24 *unit elements* of the **Hurwitz integers** form a group called the **Binary Tetrahedral Group**, denoted by **$\text{2T}$**.  It also examines $\text{2T}$'s proper subgroups and finds that they are as follows:

* one normal subgroup of order 8, isomorphic to the quaternion group, $Q_8$,
* four commutative subgroups of order 6, all isomorphic to the cyclic group, $Z_6$,
* three commutative subgroups of order 4, all isomorphic to the cyclic group, $Z_4$,
* four commutative subgroups of order 3, all isomorphic to the cyclic group, $Z_3$,
* one normal commutative subgroup of order 2, isomorphic to the cyclic group, $Z_2$.

Also, see [Wikipedia](https://en.wikipedia.org/wiki/Binary_tetrahedral_group)

```python
import finalg as fg    # The finite algebra module
from hyprat import Hu  # The Hurwitz integer class
```

## Get the 24 Hurwitz Integer Units from the `Hu` Class

```python
help(Hu.units)
```

```python
Hu.units()
```

First, shorten the strings (keys) of the half-integers shown above to be the letters 'L', 'M', ..., 'S', and their negatives.

```python
hu_units = {'1': Hu(1, 0, 0, 0),
            '-1': Hu(-1, 0, 0, 0),
            'i': Hu(0, 1, 0, 0),
            '-i': Hu(0, -1, 0, 0),
            'j': Hu(0, 0, 1, 0),
            '-j': Hu(0, 0, -1, 0),
            'k': Hu(0, 0, 0, 1),
            '-k': Hu(0, 0, 0, -1),
            'L': Hu('1/2', '1/2', '1/2', '1/2'),
            'M': Hu('1/2', '1/2', '1/2', '-1/2'),
            'N': Hu('1/2', '1/2', '-1/2', '1/2'),
            'O': Hu('1/2', '1/2', '-1/2', '-1/2'),
            'P': Hu('1/2', '-1/2', '1/2', '1/2'),
            'Q': Hu('1/2', '-1/2', '1/2', '-1/2'),
            'R': Hu('1/2', '-1/2', '-1/2', '1/2'),
            'S': Hu('1/2', '-1/2', '-1/2', '-1/2'),
            '-S': Hu('-1/2', '1/2', '1/2', '1/2'),
            '-R': Hu('-1/2', '1/2', '1/2', '-1/2'),
            '-Q': Hu('-1/2', '1/2', '-1/2', '1/2'),
            '-P': Hu('-1/2', '1/2', '-1/2', '-1/2'),
            '-O': Hu('-1/2', '-1/2', '1/2', '1/2'),
            '-N': Hu('-1/2', '-1/2', '1/2', '-1/2'),
            '-M': Hu('-1/2', '-1/2', '-1/2', '1/2'),
            '-L': Hu('-1/2', '-1/2', '-1/2', '-1/2')}
```

## Generate the Finite Algebra (Group) Defined by the Hurwitz Units

Here's the *docstring* of the finite algebra (`finalg`) method to be used. Fortunately, the `Hu`  class already supports many of the inputs required, so we can rely on the defaults provided by the method (`bin_op`, 'elem_eq`, etc.)

```python
help(fg.generate_algebra_from_element_dict)
```

```python
_2T, _, _ = fg.generate_algebra_from_element_dict(\
    hu_units,
    name="2T",
    description="Binary Tetrahedral Group")

_2T
```

```python
_2T.about()
```

## Find and Examine the Subgroups of the Binary Tetrahedral Group

```python
_2T_subs = fg.about_subalgebras(_2T)
```

### Subgroup of Order 8

The normal subgroup of order 8, found above, is isomorphic to the **Quaternion Group**.

To see this, first retrieve the Quaternion Group from `finalg`'s examples.

```python
q8 = fg.examples[13]  # Quaternion Group

q8.about()
```

Then use the method `fg.find_isomorphic_subalgebra` to go through the subgroups, ouput above by `fg.about_subalgebras`, to determine if any are isomorphic to `q8`.

```python
help(fg.find_isomorphic_subalgebra)
```

```python
iso_mapping8, _2T_8 = fg.find_isomorphic_subalgebra(q8, _2T_subs)

print(f"{iso_mapping8 = }")

_2T_8
```

So, $\text{2T}_8 \cong Q_8$

### Subgroup of Order 6 (up to Isomorphism)

The 4 subgroups of $\text{2T}$ of order 6 are all isomorphic. Since there are only 2 possible subgroups of order 6, the *cyclic group*, $Z_6$, and the *symmetric group*, $S_3$, we'll generate them and determine which one is isomorphic to the order 6 subgroups of $\text{2T}$.

```python
z6 = fg.generate_cyclic_group(6)
z6
```

```python
s3 = fg.generate_symmetric_group(3)
s3
```

```python
iso_mapping6, _2T_6 = fg.find_isomorphic_subalgebra(z6, _2T_subs)

print(f"{iso_mapping6 = }")

_2T_6
```

So, $\text{2T}_6 \cong Z_6$

### Subgroup of Order 4 (up to Isomorphism)

The 3 subgroups of $\text{2T}$ of order 4 are all isomorphic. Since there are only 2 possible subgroups of order 4, the *cyclic group*, $Z_4$, and the *Klein-4 group*, $V_4$, we'll take a chance and guess that $Z_4$ is the one isomorphic to the order 4 subgroups of $\text{2T}$.

```python
z4 = fg.generate_cyclic_group(4)
z4
```

```python
iso_mapping4, _2T_4 = fg.find_isomorphic_subalgebra(z4, _2T_subs)

print(f"{iso_mapping4 = }")

_2T_4
```

So, $\text{2T}_4 \cong Z_4$

### Subgroups of Order 3 and 2

The 3 subgroups of $\text{2T}$ of order 3 are all isomorphic and there's only a single subgroup of order 2. Since there is only one possible subgroup of order 3, and one of order 2, which must be isomorphic to $Z_3$ and $Z_2$, resp. We'll check that below.

```python
z3 = fg.generate_cyclic_group(3)
z2 = fg.generate_cyclic_group(2)
```

```python
z3
```

```python
z2
```

```python
iso_mapping3, _2T_3 = fg.find_isomorphic_subalgebra(z3, _2T_subs)

print(f"{iso_mapping3 = }")

_2T_3
```

```python
iso_mapping2, _2T_2 = fg.find_isomorphic_subalgebra(z2, _2T_subs)

print(f"{iso_mapping2 = }")

_2T_2
```

So, $\text{2T}_3 \cong Z_3$ and $\text{2T}_2 \cong Z_2$
